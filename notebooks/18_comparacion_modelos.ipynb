## Union

```python
DataUnion11 = pd.concat([DataBase12, DataEva12, DataVarios11])
DataUnion11
```

```python
DataUnion12 = (
    DataUnion11
    >> distinct()
    >> arrange(-_.map50_95)
)
DataUnion12
```

```python
(
    DataUnion12
    >> group_by(_.modelo)
    >> summarize(
        val_max = _.map50_95.max()
        , val_min = _.map50_95.min()
    )
    >> ungroup()
    >> arrange(-_.val_max)
)
```
